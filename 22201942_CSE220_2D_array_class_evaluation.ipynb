```python
#You must run this cell to print matrix and for the driver code to work
def print_matrix(m):
  row,col=m.shape
  for i in range(row):
   c=1
   print('|',end='')
   for j in range(col):
    c+=1
    if(len(str(m[i][j]))==1):
     print(' ',m[i][j],end= '  |')
     c+=6
    else:
     print(' ',m[i][j],end= ' |')
     c+=6
  print()
  print('-'*(c-col))
```

Task 1: Zigzag Walk

```python
def walk_zigzag(floor):
  even=True
  for a in range(len(floor)):
   if even==True:
    for b in range(0,len(floor[a]),2):
     if b==len(floor[a])-2:
       print(f'{floor[a][b]}',end='\n')
     else:
      print(f'{floor[a][b]}',end=' ')
    even=False
   else:
      for c in range(len(floor[a])-1, -1, -2):
        if c==1:
          print(f'{floor[a][c]}',end='\n')
        else:
          print(f'{floor[a][c]}',end=' ')
   even=True
floor = np.array([['3W' , '8B' , '4W' , '6B' , '1W' , '5B'],
                  ['3B' , '2W' , '1B' , '6W' , '3B' , '8W'],
                  ['9W' , '0B' , '7W' , '5B' , '3W' , '8B'],
                  ['2B' , '1W' , '3B' , '6W' , '0B' , '4W'],
                  ['1W' , '4B' , '2W' , '8B' , '6W' , '6B']])
print_matrix(floor)
print()
print('Walking Sequence:')
walk_zigzag(floor)
#This should print
# 3W 4W 1W
# 8W 6W 2W
# 9W 7W 3W
# 4W 6W 1W
# 1W 2W 6W
```

Task 2: Landscape Screen

```python
def landscape(m):
  new_row_no, new_col_no = len(m[3]), len(m)  # swapping row and column numbers
  transposed_arr = np.zeros((new_row_no, new_col_no), dtype = int)
  for c in range(new_col_no): # looping through columns
    for r in range(new_row_no): # looping through rows
      transposed_arr[r][c] = m[c][r]
  return transposed_arr
m = np.random.randint(15,size=(4,3)) #This creates an np array of 4 x 3 where the values are random numbers from [0,15) range
print_matrix(m)
print()
print_matrix(landscape(m))
```

Task 3: Seating Arrangement

```python
def arrangement(m, diff):
  r,c=len(m),len(m) # as square shaped
  for row in range(r):
    for column in range(c):
      if row!= column and abs(m[row][column]-m[column][row])!=diff:
        return False
  return True
m = np.array([[7,13,9,14],[12,8,15,11],[10,17,3,2],[15,10,1,4]])
print_matrix(m)
print()
returned_value = arrangement(m,1)
print(returned_value) # This should print False
unittest.output_test(returned_value, False)

print()

n = np.array([[7,13,9,14],[12,8,16,11],[10,17,3,2],[15,10,1,4]])
print_matrix(n)
print()
returned_value = arrangement(n,1)
print(returned_value) # This should print True
unittest.output_test(returned_value, True)
```

Task 4: Chess Piece

```python
def show_knight_move(knight):
  pos = np.zeros((8, 8), dtype = int)
  pos[knight[0]][knight[1]], pos_indicator = 33, 3
  possible_knight_move = np.array([(-2, 1), (-2, -1), (2, 1), (2, -1), (1, 2), (-1, 2), (1, -2), (-1, -2)]) #  8 positions of knight
  for idx in range(len(possible_knight_move)):
    var_x, var_y = possible_knight_move[idx][0] + knight[0], possible_knight_move[idx][1] + knight[1]
    if var_x in range(8) and var_y in range(8):
      pos[var_x][var_y] = pos_indicator
  return pos
knight = (3, 4)
chess_board = show_knight_move(knight)
print_matrix(chess_board)
```

```python
def topple(knight, rook):
  chess = show_knight_move(knight)
  chess[knight[0]][knight[1]], chess[rook[0]][rook[1]] = 33, 5
  knight_moves=np.array([(-2, 1), (-2, -1), (2, 1), (2, -1), (1, 2), (-1, 2), (1, -2), (-1, -2)])#possible 8 positions
  if rook[0]==knight[0] or rook[1] == knight[1]: #checking if rook can kill knight
   return chess,'Rook can kill'
  else:
   for i in range(len(knight_moves)):
    x, y=knight_moves[i][0]+knight[0],knight_moves[i][1]+knight[1]
    if rook[0] == x and rook[1] == y:
     return chess, 'Knight can kill'
  return chess, 'Cannot Kill'
knight = (5, 6)
rook = (4, 7)
chess_board, ans = topple(knight,rook)
print_matrix(chess_board)
print(ans)#This should print Rook can kill
```
