```python
class Node:
  def __init__(self, element, next):
    self.element = element
    self.next = next
class Stack:
  def __init__(self):
    self.top = None # an empty stack
  def push(self, element):
    if self.top == None:
      self.top = Node(element, None)
    else:
      node = Node(element, None)
      node.next = self.top
      self.top = node
  def pop(self):
    if self.top == None:
      return -1  #this is error
    else:
      popped_node = self.top
      self.top = popped_node.next
      popped_node.next = None
      return popped_node.element
  def peek(self):
    if self.top!= None:
      return self.top.element
    return -1
  def empty(self):
    return self.top == None
def mergeIntervals(set_of_intervals):
  res_stack = Stack()
  res_stack.push(set_of_intervals[0])
  for i in range(1, len(set_of_intervals)):
    current_start, current_end=res_stack.peek()
    if set_of_intervals[i][0]<=current_end: #overlap
      if current_end<set_of_intervals[i][1]:
       res_stack.pop()               #pop current
       res_stack.push((current_start,set_of_intervals[i][1]))
    else:
      res_stack.push(set_of_intervals[i])
  while res_stack.empty()!= True:
   print(res_stack.pop())
mergeIntervals([(1, 5), (2, 3), (4, 6), (7, 8), (8, 10), (12, 15)])
```

```python
import numpy as np
img = [[1,1,0],[1,0,1],[0,0,0]]
def flipAndInvertImage(image):
  rows=len(image)
  new_img=np.zeros((rows,rows), dtype = int)
  for i in range(rows):
    for j in range(rows):
      new_img[i][rows-1-j]=image[i][j]
  for i in range(rows):
   for j in range(rows):
    if new_img[i][j]== 0:
     new_img[i][j]= 1
    else:
     new_img[i][j]= 0
  return new_img
print(flipAndInvertImage(img))
```

```python
def rotateLeft(array):

  l = len(array)

  temp = array[0]

  for i in range(0, l-1):

    array[i] =array[i+1]

  array[l-1] = temp
  return array

def printTime(hours, minutes, jump):

  change_hours = jump // 60
  change_minutes = jump % 60

  for h in range(change_hours):
    hours = rotateLeft(hours)

  for m in range(change_minutes):
    minutes = rotateLeft(minutes)

  if minutes[0] >= 10:
    time = str(hours[0]) + ":" + str(minutes[0])
  else:
    time = str(hours[0]) + ":0" + str(minutes[0])


  return time

hours = np.arange(1, 13)
minutes = np.arange(60)

printTime(hours, minutes, 719)
```

```python
import numpy as np

def shiftRight(array):
  l=len(array)
  for i in range(l-1, 0, -1):
    array[i]=array[i-1]
  array[0]=0
  return array
print(shiftRight(np.array([1,2,3,4])))
def shiftRightbyK(array,K):
  for i in range(K):
    array=shiftRight(array)
  return array
print(shiftRightbyK(np.array([1,2,3,4]),3))
```

```python
def shiftLeft(array):
  l=len(array)
  for i in range(0, l-1):
    array[i]=array[i+1]
  array[l-1]=0
  return array
print(shiftLeft(np.array([1,2,3,4])))
def removeByIndex(array, index):
  l=len(array)
  for i in range(index, l-1):
    array[i]=array[i+1]
  array[l-1]=0
  return array
print(removeByIndex(np.array([1,2,3,4]), 2))
def removeAllOccurancesByValue(array, value):
  l = len(array)
  for i in range(l):
   while array[i] == value:
    array = removeByIndex(array, i)
  return array
print(removeAllOccurancesByValue(np.array([1,2,2,3,4,2,2,2]), 2))
```
